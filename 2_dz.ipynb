**Задача 1**. (11 баллов)

В известном магазине техники представлено очень много экземпляров различной бытовой техники двух видов: с возможностью подключения к системе "Умный дом" и без такой возможности. Причем, известно, что эти два вида техники представлены в пропорции 1 : 3 соответственно. Случайным образом выбрали 7 экземпляров техники для розыгрыша среди участниц конкурса "Мисс Рукоделье - 2024".  Какова вероятность, что среди них не более двух экземпляров с возможностью подключения к системе "Умный дом"?

Подробно распишите решение и обоснуйте выбор распределения для моделирования сюжета задачи. Ответ округлите до трёх знаков.


```python
from math import comb as C
from scipy.stats import binom

'''
Биноминальное распределление:
сами испытания не зависят друг друга, исходов испытания два, известны частота успеха и неуспеха для генеральной совокупности.
'''
p = 0.25 # 1 (1 + 3)
q = 0.75 # 3 / (1 + 3)
n = 7
'''
не более двух экземпляров -- это 0, 1, 2
Воспользуемся формулой биноминального распределдения для определения вероятности встретить k ууспехов в серии n = 7 испытаний
C(n, k) тут число сочетаний из n по k (из 7 по 0, по 1, по 2)
'''
result = C(7, 0) * p**0 * q**7 + C(7, 1) * p**1 * q**6 + C(7, 2) * p**2 * q**5
print('Ответ: ', result)
#distribution = binom(n, p)
#python_result = distribution.pmf(0) + distribution.pmf(1) + distribution.pmf(2)
#print(python_result)
```

Дан закон распределения количества публикаций в которой группе телеграм-каналов:

$$
\begin{array}{|c|c|c|c|c|c|c|}
\hline X \text { (число публикаций) } & 0 & 1 & 2 & 3 & 4 & 5 \\
\hline \mathrm{P} \text { (вероятность) } & ? & 0,3 & 0,3 & 0,2 & 0,1 & 0,1 \\
\hline
\end{array}
$$

Найдите пропущенную вероятность для X = 0, математическое ожидание и дисперсию. Решите задачу "вручную" и с помощью python.

```python
import numpy as np
'''
Сумма вероятностей должна быть 1
x + 0.3 + 0.3+ 0.2 + 0.1 + 0.1 = 1
x = 0
'''
print('P[X=0] = 0')
# Вычисление математического ожидания
E_manual = 0 * 0 + 1 * 0.3 + 2* 0.3 + 3* 0.2 + 4* 0.1 + 5* 0.1
print('Математическое ожидание, вычисленное руками: ', E_manual)
'''
Вычислим дисперсию как разность между математическим ожиданием квадрата и квадратом математического ожидания
E[X**2] - (E[X])**2
'''
D_manual = 0 * 0 + 1 * 0.3 + 4* 0.3 + 9* 0.2 + 16* 0.1 + 25* 0.1 - E_manual**2
print('Дисперсия, вычисленная руками: ', D_manual)

print('\nРешение при помощи numpy:\n')
X_vals = np.array([0, 1, 2, 3, 4, 5])
P_vals = np.array([0, 0.3, 0.3, 0.2, 0.1, 0.1])
E = X_vals @ P_vals
D = X_vals**2 @ P_vals - E**2
print('Математическое ожидание: ', E)
print('Дисперсия: ', D)

```

**Задача 3** (25 баллов)


Компания приобрела 10 000 новых ноутбуков. Вероятность поломки одного ноутбука в течение месяца равна 0,0002. Предполагается, что поломки отдельных ноутбуков происходят независимо друг от друга.

**Вопрос 1**. Найдите вероятность того, что за месяц сломается ровно 1 ноутбук.
Решение объясните. (5 баллов)


**Вопрос 2**. Компания планирует ввести программу расширенной гарантии, по которой она будет бесплатно ремонтировать или заменять ноутбуки, сломавшиеся в течение месяца. Если общее количество поломок за месяц превысит 5 ноутбуков, компания понесет значительные дополнительные расходы.

а) Найдите вероятность того, что за месяц сломается более 5 ноутбуков.Решение объясните (10 баллов)

б) Если стоимость ремонта или замены одного ноутбука составляет 500 долларов, вычислите математическое ожидание общих расходов компании на поломки за один месяц. (10 баллов)

```python
'''
ВОПРОС 1
События независимы, известна вероятность успеха испытания (под успехом понимается поломка)
Имеем делом с биноминальным распределением
'''
p = 0.0002
q = 1 - p
n = 10000
result = C(n, 1) * p**1 * q**(n-1)
print(result)
#print(binom.pmf(1, n, p))
```

```python
'''
Вопрос 2.a
События "сломается более 5 ноутбуков" и сломается "не более 5 ноутбуков" несовместны и образуют полную группу (сумма их вероятностей 1)
ответ 1 - P(X<=5)
'''
P_X_0 = C(n, 0) * p**0 * q**n
P_X_1 = C(n, 1) * p**1 * q**(n-1)
P_X_2 = C(n, 2) * p**2 * q**(n-2)
P_X_3 = C(n, 3) * p**3 * q**(n-3)
P_X_4 = C(n, 4) * p**4 * q**(n-4)
P_X_5 = C(n, 5) * p**5 * q**(n-5)

result = 1 - (P_X_0 + P_X_1 + P_X_2 + P_X_3 + P_X_4 + P_X_5)
print(result)
#print(1 - binom.cdf(5, n, p))
```

```python
'''
Вопрос 2.б
Нас интересует E[loss], где loss = 500 * X и X - количество сломанных ноутбуков. X --случайная величина,
распределённая по биноминальному закону,
но 500 -- константа, а потому E[loss] = 500 * E[X].
Для биноминального распределения E[X]  = n * p
'''
result = 500 * n * p
print(result)
```
