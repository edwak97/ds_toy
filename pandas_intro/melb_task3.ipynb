```python
import pandas as pd
import numpy as np

melb_data = pd.read_csv('data/melb_data_tuned.csv', sep=';')

melb_data.head()


```

```python
# a copy
melb_df = melb_data.copy()
```

```python
# let's drop index AND Coordinates columns
melb_df = melb_data.drop(['index', 'Coordinates'], axis=1)
melb_df.head()
```

```python
# let's sum Rooms, Bedroom,Bathroom
total_rooms = melb_df['Rooms'] + melb_df['Bedroom'] + melb_df['Bathroom']
total_rooms
```

```python
# let's build new column asaratio of BuildingArea and total_rooms
melb_df['MeanRoomsArea'] = melb_df['BuildingArea'] / total_rooms
melb_df.MeanRoomsArea
```

```python
# area ratio
diff_area = melb_df['BuildingArea'] - melb_df['Landsize']
sum_area = melb_df['BuildingArea'] + melb_df['Landsize']
melb_df['AreaRatio'] = diff_area / sum_area
melb_df.AreaRatio
```

```python
countries_df = pd.DataFrame({
    'country': ['Англия', 'Канада', 'США', 'Россия', 'Украина', 'Беларусь', 'Казахстан'],
    'population': [56.29, 38.05, 322.28, 146.24, 45.5, 9.5, 17.04],
    'area': [133396, 9984670, 9826630, 17125191, 603628, 207600, 2724902]
})
countries_df['density'] = countries_df['population'] / countries_df['area']
round(countries_df.density.mean()*1e6,2)
```
