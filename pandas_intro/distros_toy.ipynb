```python
import numpy as np
from scipy.stats import randint, binom

x = np.arange(1, 7)
dist = randint(1, 7)
print(dist.pmf(x))

#Равсчёт кумулятивной функции распределения
print(dist.cdf(x))
```

```python
from scipy.stats.distributions import poisson

# в колл-центр звонят 4.5 раза за 5 минут
# с какой вероятносмтью будет совершено 3 звонка?
# для какого наименьшего количества звонков cfd будет не менее 0.9?
best_count = 0
for i in range(100):
    if poisson.cdf(i, 4.5) >= 0.9:
        best_count = i
        break
print(best_count)
print('Этого количества операторов хватило бы для 90% случаев')
```

```python
'''
Пассажиры прибывают на вокзал со средней скоростью человека в минуту.

Если количество пассажиров, приезжающих на вокзал,
подчиняется распределению Пуассона, какова приблизительная вероятность того,
что 16 пассажиров приедут на вокзал в конкретный четырёхминутный период?
Округлите ответ до трёх знаков после точки-разделителя.
'''

from scipy.stats import poisson, expon

lambda_ = 4*4
k = 16

prob = poisson.pmf(k, lambda_)
print(round(prob, 3))

# холодильник служит 10 лет и продолжительностьь службы распределена экспоненциально
# с какой вероятносттью он будет служить более 7 лет?
prob = 1 - expon.cdf(7, scale=10)
print(round(prob, 3))

#от 9 до 11
prob = expon.cdf(11, scale=10) - expon.cdf(9, scale=10)
print(round(prob, 3))
```
