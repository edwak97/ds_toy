```python
import numpy as np
from scipy.optimize import minimize
from scipy.optimize import least_squares
import pandas as pd

np.array([1, 2, 3])
```

```python
def func_rosen(x):
    r = np.sum(100*(x[1:]-x[:-1]**2)**2 + (1-x[:-1])**2, axis=0)
    return r
x1 = np.array([2.4, 1.5, 2.1, 0.7, 1.1])
result_point = minimize(func_rosen, x1, method='powell')
print(result_point.x)
print(result_point.fun)
print(result_point.success)
print(result_point.message)
```

```python
#Симплекс-метод
result_point = minimize(func_rosen, x1, method='nelder-mead')
print(result_point.x)
print(result_point.fun)
print(result_point.success)
```

```python
#метоод наименьших квадратов
result_point = least_squares(func_rosen, x1)
print(np.round(result_point.x, 3))
print(result_point.fun)
```

```python
## Решение следующей точки градиентного спуска для точки (x, y, z) = (0, 0, 0)
# функции f и alkpha = 0.25
from sympy import Symbol, diff
x = Symbol('x')
y = Symbol('y')
z = Symbol('z')
f = 2*x**2 -4*x*z + 4*y**2 - 8*y*z+9*z**2+4*x+8*y-20*z
grad = [diff(f, var) for var in [x, y, z]]
print(grad)
```

```python
alpha = 0.25
alpha_grad = [alpha*grad_component.subs({x: 0, y: 0, z: 0}).evalf() for grad_component in grad]
print(alpha_grad)
```

```python
## Реализуем градиентный спуск ручками
def func(x, y):
    return 2*x**2 - 4*x*y + y**4 + 2
def derivative(x, y):
    return 4*x - 4*y, -4*x + 4*y**3

def getMinimumByGradientDescent(coords_start, func, func_derivative, learning_rate=0.01, steps=10**5, limit=1e-3):
    f_val=func(*coords_start)
    delta = 1
    while steps > -1 and delta >limit:
        tup_dxdy = func_derivative(*coords_start)
        new_coord = len(coords_start) * [None]
        for i in range(len(coords_start)):
            new_coord[i] = coords_start[i] - learning_rate*tup_dxdy[i]
        new_f_val=func(*new_coord)
        delta = abs(new_f_val - f_val)
        steps -=1
        coords_start = new_coord
        f_val = new_f_val
    print(x, y, f_val, steps, delta)
getMinimumByGradientDescent((4, 7), func=func, func_derivative=derivative)
def fun2(x, y, z, /):
    return x**3-2*x**2+y**2+z**2-2*x*y+x*z-y*z+3*z
def fun2_derivative(x, y, z, /):
    return 3*x**2-4*x-2*y+z, 2*y-2*x-z, 2*z + x - y + 3

getMinimumByGradientDescent((4, 7, 8), func=fun2, func_derivative=fun2_derivative)
```

```python
import sympy as sp
x = sp.Symbol('x')
f_x = -2 * x**3 - 3*x**2 + 12*x - 4
#найдём производную и приравняем к 0
f_x_prime = sp.diff(f_x, x)
sp.solve(f_x_prime, x)
print(f_x.subs(x, -2))
print(f_x.subs(x, 1))

#Состиавление матрицы Гессе для f(x, y) = e**x + 5y**3 *x
y = sp.Symbol('y')
fxy = sp.exp(x) + 5*y**3*x
f_x_prime = sp.diff(fxy, x)
f_y_prime = sp.diff(fxy, y)
f_x_prime_0_0 = sp.diff(f_x_prime, x)
f_y_prime_1_1 = sp.diff(f_y_prime, y)
f_y_prime_mix = sp.diff(f_y_prime, x)
#Hesse matrix
print([[f_x_prime_0_0, f_y_prime_mix], [f_y_prime_mix, f_y_prime_1_1]])
```
