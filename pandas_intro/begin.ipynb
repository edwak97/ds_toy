```python
import pandas as pd

print(pd.__name__)
```

```python
countries = pd.Series(
    data = ['England', 'France', 'Germany', 'Spain', 'United States'],
    index = ['UK', 'FR', 'DE', 'ES', 'US'],
    name = 'countries'
)

display(countries)
```

```python
print(countries.loc['US'])
print('You get series for this!')
print(countries.loc[['US','UK']])
```

DataFrame -- это таблица с данными
Series -- это строка или столбец

```python
countries_df = pd.DataFrame({
    'country': ['Англия', 'Канада', 'США', 'Россия', 'Украина', 'Беларусь', 'Казахстан'],
    'population': [56.29, 38.05, 322.28, 146.24, 45.5, 9.5, 17.04],
    'area': [133396, 9984670, 9826630, 17125191, 603628, 207600, 2724902],
}, index=['UK', 'CA', 'US', 'RU', 'UA', 'BY', 'KZ'])
display(countries_df)

countries_df = pd.DataFrame(
    data = [['Англия', 56.29, 133396],
            ['Канада', 38.05, 9984670],
            ['США', 322.28, 9826630],
            ['Россия', 146.24, 17125191],
            ['Украина', 45.5, 603628],
            ['Беларусь;', 9.5, 207600],
            ['Казахстан', 17.04, 2724902]],
    columns = ['country', 'population', 'area'],
    index = ['UK', 'CA', 'US', 'RU', 'UA', 'BY', 'KZ']
)
display(countries_df)
```

О доступе к строкам и столбцам

```python
print('Площадь России:')
countries_df.loc['RU', 'area'] #countries_df.loc['RU'].area

print('Население и площадь России')
countries_df.loc['RU', ['population', 'area']]
```

Запишем DataFrame в файл

И прочитаем

```python
countries_df.to_csv('data/countries.csv', index=False, sep = ';')

data = pd.read_csv('data/countries.csv', sep = ';')
display(data)
```
