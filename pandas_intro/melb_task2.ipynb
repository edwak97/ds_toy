```python
import pandas as pd
import numpy as np

melb_data = pd.read_csv('data/melb_data_tuned.csv', sep=';')

rate = 0.12
income = melb_data['Price'].sum() * rate
# total income of estate agencies:
round(income,2)


```

```python
#Отличие медианнорй пшлощади от среднего
landsize_median = melb_data['Landsize'].median()
landsize_mean = melb_data['Landsize'].mean()
abs(landsize_median - landsize_mean)/landsize_mean
```

```python
# наиболее распространённое название района
melb_data['Regionname'].mode()
```

```python
#максимальтное количество домов на продажу
melb_data['Propertycount'].max()
```

```python
#разброс расстояния от центра города до объекта недвижимости
melb_data['Distance'].std()
```

```python
#Отличие медианной площади здания от среднего значения
landsize_median = melb_data['BuildingArea'].median()
landsize_mean = melb_data['BuildingArea'].mean()
round(abs(landsize_median - landsize_mean)/landsize_mean * 100, 0)
```
