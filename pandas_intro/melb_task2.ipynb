```python
import pandas as pd
import numpy as np

melb_data = pd.read_csv('data/melb_data_tuned.csv', sep=';')

rate = 0.12
income = melb_data['Price'].sum() * rate
# total income of estate agencies:
round(income,2)


```

```python
#Отличие медианнорй пшлощади от среднего
landsize_median = melb_data['Landsize'].median()
landsize_mean = melb_data['Landsize'].mean()
abs(landsize_median - landsize_mean)/landsize_mean
```

```python
# наиболее распространённое название района
melb_data['Regionname'].mode()
```

```python
#максимальтное количество домов на продажу
melb_data['Propertycount'].max()
```

```python
#разброс расстояния от центра города до объекта недвижимости
melb_data['Distance'].std()
```

```python
#Отличие медианной площади здания от среднего значения
landsize_median = melb_data['BuildingArea'].median()
landsize_mean = melb_data['BuildingArea'].mean()
round(abs(landsize_median - landsize_mean)/landsize_mean * 100, 0)
```

```python
#СКОЛЬКО СПАЛЕН ЧАЩЕ ВСЕГО ВСТРЕЧАЕТСЯ?
melb_data['Bedroom'].mode()
```

О фильтрации

```python
# Количество зданий с тремя комнатами
melb_data[melb_data['Rooms'] == 3].shape[0]
```

```python
#дома с ценой менее 300К и (с числом комнат 3 ИЛИ площадью более 100)
melb_data[
    (melb_data['Price'] < 300000) & 
    ((melb_data['Rooms'] == 3) | (melb_data['BuildingArea'] > 100))
].shape[0]
```

```python
#Найдём медианную площадь у объектов, чья цена выше средней

melb_data[melb_data['Price'] > melb_data['Price'].mean()]['BuildingArea'].median()
```

```python
# У скольких объектов отсустствуют ванные комнаты?
melb_data[melb_data['Bathroom'] == 0].shape[0]
```

```python
#Сколько в таблице melb_data объектов недвижимости,
# которые были проданы риелторской компанией Nelson и стоимость которых составила больше 3 миллионов?

melb_data[
    (melb_data['SellerG'] == 'Nelson') & 
    (melb_data['Price'] > 3e6)
].shape[0]
```

```python
# Минимальная стоимость участка без здания
melb_data[melb_data['BuildingArea'] == 0]['Price'].min()
```

```python
# Какова средняя цена объектов недвижимости в таблице melb_data
# с ценой менее одного миллиона,
# в которых либо количество комнат больше пяти, либо здание моложе 2015 года?
res = melb_data[
    (melb_data['Price'] < 1e6) &
    (
        (melb_data['Rooms'] > 5) | (melb_data['YearBuilt'] < 2015)
    )
]['Price'].mean()
round(res,0)
```

```python
# В какомрайоне чаще всего продаются здания типа h сценой меньще 3 милллионов
melb_data[(melb_data['Price'] < 3e6) & (melb_data['Type'] == 'h')]['Regionname'].mode()
```
