**Задача 1** (14 баллов)

Найдите экстремумы функции:

$$
f(x, y, z)=2 x^3+2 x y+2 x z+y^2+z^2+2 y-8 .
$$

Распишите подробное решение.

```python
# найдём частные производные
'''
dfdx = 6x**2 + 2*y + 2*z
dfdy = 2*x + 2*y + 2
dfdz = 2*x + 2*z
'''
# некобходимо,чтобы частные производные обращались в 0. Найдём соответствующие значения, решив систему уравнений:
'''
dfdx = 0
dfdy = 0
dfdz = 0

выразим y во втором уранении через x, z в третьем уранении через x,
заменим соответствующие переменные в первом уравнении
y = -x - 1
z = -x

значит
6x**2 + 2*(-x-1) + 2*(-x) = 0
два корня: -1/3, 1;
стало быть, есть два кандидата:
x = -1/3, y = -2/3, z = 1/3
x = 1, y = -2, z = -1
'''
#Проверим решение при помощи Sympy
from sympy import *
x, y, z = symbols('x y z')
f = 2*x**3 + 2 *x*y + 2*x*z +y**2 + z**2 + 2*y - 8
dfdx, dfdy, dfdz = diff(f, x), diff(f, y), diff(f, z)
solution = solve([dfdx, dfdy, dfdz], [x, y, z])
print(solution)
#анализ гессиана,поиск производных второго порядка
'''
dfdxdx = 12x
dfdxdy = 2
dfdxdz = 2

dfdydy = 2
dfdyxy = 2
dfdydz = 0

dfdzdx = 2
dfdzdy = 0
dfdzdz = 2

Гессиан имеет вид:
[[12x, 2, 2], 
[2,    2, 0], 
[2,    0, 2]]

Расммотрим сначала первую точку (x = -1/3, y = -2/3, z = 1/3)
минор1 = -4, минор2= (-4 * 2 - 2 * 2) = -12;
первый угловой минор отрицательный, но второй тоже отрицатетельный -- имеем дело с седловой точкой.

Рассмотрим сначала вторую точку (x = 1, y = -2, z = -1)
минор1 = 12, минор2 = (12*2 - 2 * 2) = 20;
третий получим разложением третьей строки: 2 * (2*0-4) + 0 + 2 * (12*2-2*2) = -8 + 40 = 32
все три минора положительны, имеем дело с локальным минимумом
'''
#проверим вычисления при помощи sympy, numpy
import numpy as np
from numpy.linalg import eig

def get_hessian(f, x_val, y_val, z_val):
    '''
    Возвращает гессиан функции f в точке (x_val, y_val, z_val)
    '''
    x, y, z = symbols('x y z')
    args = {x: x_val, y: y_val, z: z_val}
    dfdxdx, dfdydy, dfdzdz = diff(f, x, x), diff(f, y, y), diff(f, z, z)
    dfdxdy, dfdxdz = diff(f, x, y), diff(f, x, z)
    dfdydz = diff(f, y, z)
    return np.array([
        [dfdxdx.subs(args), dfdxdy.subs(args), dfdxdz.subs(args)],
        [dfdxdy.subs(args), dfdydy.subs(args), dfdydz.subs(args)],
        [dfdxdz.subs(args), dfdydz.subs(args), dfdzdz.subs(args)]
    ]).astype(np.float64)
hessian1 = get_hessian(f, -1/3, -2/3, 1/3)
hessian2 = get_hessian(f, 1, -2, -1)
print('Собственные числа первого гессиана (знакопеременные,седловая точка):', eig(hessian1)[0] )   
print('Собственные числа второго гессиана (все собственные числа положительные, локальный минимум):', eig(hessian2)[0])
print('Таким образом, исследуемая функция имеет одинэкстремум: локальный минимум в точке 1, -2, 1')
```

**Задача 2** (25 баллов)

Найдите условные экстремумы функции:

$$
f(x, y)=4 x+8 y, y^2-2 x y+5=0
$$

Вычислите результат самостоятельно (вручную) и с помощью Python. Сравните результаты.

```python
#Найдём элементы функции Лагранжа
'''
f(x, y) = 4*x + 8*y
g(x, y) = y**2 - 2 * x * y + 5
'''

#Составим функцию Лагранжа
'''
L(x, y, lambd) = f(x, y) + lambd * g(x, y)
L(x, y, lambd) = 4*x+8*y+lambd*(y**2-2*x*y+5)
'''

#Найдём частные производные функции Лагранжа по всем трём переменным
'''
dLdx = 4 - 2 * lambd*y
dLdy = 8 + 2 * lambd*y - 2 *lambd*x
dLdlambd = y**2 - 2 * x * y + 5
'''

#Приравняем к 0 и решим систему уравнений
'''
lambd * y - 2 = 0
lambd * y - lambd * x + 4 = 0
y**2 - 2 * x * y + 5 = 0
Решением этой системы будет:
x   y   lambd
-3 -1 -2
 3, 1, 2
'''
#Найдём значения целевой функции в соответствующих точках
'''
f(-3, -1) = 4 * (-3) + 8 * (-1) = -20
f(3, 1) = 4 * 3 + 8 * 1 = 20
'''

# Посчитаем при помощи Python
import sympy as sp
x, y, l = sp.symbols('x y l')
f = 4*x + 8*y
print('Целевая функция:', f)
g = y**2 - 2 * x * y + 5
L = f + l * g
print('Функция Лагранжа:', L)

sol = sp.solve([sp.diff(L, x), sp.diff(L, y), sp.diff(L, l)], [x, y, l])
print('Решение. Стационарные точки:', ", ".join(map(str, sol)))
sol_xy = [(s[0], s[1]) for s in sol]

f_vals = [f.subs(x, s[0]).subs(y, s[1]) for s in sol_xy]
print('Значения целевой функции в стационарных точках:', ", ".join(map(str, f_vals)))
```

**Задача 3** (30 баллов)

Вам предложены данные с информацией о успеваемости студентов:

Независимые переменные:
* Hours Studied: Общее количество часов, потраченных на учебу каждым студентом.
* Previous Scores: Баллы, полученные студентами на предыдущих экзаменах.
* Sleep Hours: Среднее количество часов сна студента в сутки.
* Sample Question Papers Practiced: Количество пробных экзаменационных работ, с которыми студент занимался.
  
Целевая переменная:
* Performance Index: Показатель общей успеваемости каждого студента. Индекс успеваемости отражает академическую успеваемость студента и округляется до ближайшего целого числа. Индекс варьируется от 10 до 100, при этом более высокие значения свидетельствуют о более высокой успеваемости.

**Решите задачу линейной регрессии, реализовав градиентный спуск самостоятельно, не используя готовое решение из какой-либо библиотеки.**

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split

df = pd.read_csv('skillfactory_asset.txt')
#заменим yes/no на 1/0
df['Extracurricular Activities'] = df['Extracurricular Activities'].map({'Yes': 1, 'No': 0})
# print(df.head())
# df.info()
#y -- целевая переменная
y = df['Performance Index'].values
n = y.shape[0]
y = y.reshape(n, 1)

X = df.drop('Performance Index', axis=1).values
X_std = (X - X.mean(axis=0)) / X.std(axis=0)
X_std = np.append(np.ones((n, 1)), X_std, axis=1)
# print(X.shape)
# print(y.shape)
# params = np.zeros((X.shape[1], 1))
# print(params.shape)


def cost_f(X, y,params):
    y_pred = np.dot(X, params)
    return 1/(len(y)) * np.sum((y_pred - y)**2)

def grad_d(f, X, y ,par, learning_rate, max_iter = 30000, eps = 1e-4):
    prev_f = f(X, y, par)
    passed_iters = 0
    delta = 10
    while(delta > eps and passed_iters < max_iter):
        y_pred = np.dot(X, par)
        der =  np.dot(X.T, y_pred - y)/len(y)
        par -= der * learning_rate
        cur_f = f(X, y, par)
        delta = abs(prev_f - cur_f)
        prev_f = cur_f
        passed_iters += 1
    return par, passed_iters, delta

x_train, x_test, y_train, y_test = train_test_split(X_std, y, test_size=0.3, random_state=42)

params, iters, delta = grad_d(cost_f, x_train, y_train, np.zeros((x_train.shape[1], 1)), learning_rate=0.01)
print('Параметры (градиентный спуск):\n', params)
print(f'Сошлось за {iters} итераций, последнее изменение функции потерь: {delta:.3e}')

train_cost = cost_f(x_train, y_train, params)
print(f'Значение функции потерь (MSE) на train по результатам градиентного спуска: {train_cost:.4f}')
y_pred = np.dot(x_test, params)
test_cost = np.mean((y_pred - y_test)**2)
print(f'Значение функции потерь на test (сравнение предсказания с эталонным данными вне обучения):  {test_cost:.4f}')

# через существуцющие инструменты
print('\nС помощью стандартных инструментов:')
import statsmodels.api as sm
mod = sm.OLS(y_train, x_train)
res = mod.fit()
print(res.params)
print(f'Значение функции потерь (MSE) на train по результатам работы statmodels: {res.mse_resid:.4f}')
pred = res.predict(x_test).reshape(-1, 1)
test_cost2 = np.mean((pred - y_test)**2)
print(f'Значение функции потерь (MSE) на test:  {test_cost2:.4f}')



```
