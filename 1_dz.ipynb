**Задача 1** (14 баллов)

Найдите экстремумы функции:

$$
f(x, y, z)=2 x^3+2 x y+2 x z+y^2+z^2+2 y-8 .
$$

Распишите подробное решение.

```python
# найдём частные производные
'''
dfdx = 6x**2 + 2*y + 2*z
dfdy = 2*x + 2*y + 2
dfdz = 2*x + 2*z
'''
# некобходимо,чтобы частные производные обращались в 0. Найдём соответствующие значения, решив систему уравнений:
'''
dfdx = 0
dfdy = 0
dfdz = 0

выразим y во втором уранении через x, z в третьем уранении через x,
заменим соответствующие переменные в первом уравнении
y = -x - 1
z = -x

значит
6x**2 + 2*(-x-1) + 2*(-x) = 0
два корня: -1/3, 1;
стало быть, есть два кандидата:
x = -1/3, y = -2/3, z = 1/3
x = 1, y = -2, z = -1
'''
#Проверим решение при помощи Sympy
from sympy import *
x, y, z = symbols('x y z')
f = 2*x**3 + 2 *x*y + 2*x*z +y**2 + z**2 + 2*y - 8
dfdx, dfdy, dfdz = diff(f, x), diff(f, y), diff(f, z)
solution = solve([dfdx, dfdy, dfdz], [x, y, z])
print(solution)
#анализ гессиана,поиск производных второго порядка
'''
dfdxdx = 12x
dfdxdy = 2
dfdxdz = 2

dfdydy = 2
dfdyxy = 2
dfdydz = 0

dfdzdx = 2
dfdzdy = 0
dfdzdz = 2

Гессиан имеет вид:
[[12x, 2, 2], 
[2,    2, 0], 
[2,    0, 2]]

Расммотрим сначала первую точку (x = -1/3, y = -2/3, z = 1/3)
минор1 = -4, минор2= (-4 * 2 - 2 * 2) = -12;
первый угловой минор отрицательный, но второй тоже отрицатетельный -- имеем дело с седловой точкой.

Рассмотрим сначала вторую точку (x = 1, y = -2, z = -1)
минор1 = 12, минор2 = (12*2 - 2 * 2) = 20;
третий получим разложением третьей строки: 2 * (2*0-4) + 0 + 2 * (12*2-2*2) = -8 + 40 = 32
все три минора положительны, имеем дело с локальным минимумом
'''
#проверим вычисления при помощи sympy, numpy
import numpy as np
from numpy.linalg import eig

def get_hessian(f, x_val, y_val, z_val):
    '''
    Возвращает гессиан функции f в точке (x_val, y_val, z_val)
    '''
    x, y, z = symbols('x y z')
    args = {x: x_val, y: y_val, z: z_val}
    dfdxdx, dfdydy, dfdzdz = diff(f, x, x), diff(f, y, y), diff(f, z, z)
    dfdxdy, dfdxdz = diff(f, x, y), diff(f, x, z)
    dfdydz = diff(f, y, z)
    return np.array([
        [dfdxdx.subs(args), dfdxdy.subs(args), dfdxdz.subs(args)],
        [dfdxdy.subs(args), dfdydy.subs(args), dfdydz.subs(args)],
        [dfdxdz.subs(args), dfdydz.subs(args), dfdzdz.subs(args)]
    ]).astype(np.float64)
hessian1 = get_hessian(f, -1/3, -2/3, 1/3)
hessian2 = get_hessian(f, 1, -2, -1)
print('Собственные числа первого гессиана (знакопеременные,седловая точка):', eig(hessian1)[0] )   
print('Собственные числа второго гессиана (все собственные числа положительные, локальный минимум):', eig(hessian2)[0])
print('Таким образом, исследуемая функция имеет одинэкстремум: локальный минимум в точке 1, -2, 1')
```
