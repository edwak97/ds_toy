```python
import numpy as np
print('That\'s a random value:', np.random.rand())
print('A matrix 4x3 of them:\n', np.random.rand(4, 3))

res = np.zeros(10000, dtype=np.float32)
for i in range(10000):
    temp = np.random.randint(low=100, high=201, size=100)
    res[i] = len(temp[temp >= 150])
print(res.mean())

```

```python
arr = np.arange(10)
print(arr)
np.random.shuffle(arr)
print(arr)

#то же самое
arr = np.random.permutation(10)
print(arr)

#новый массив
arr2 = np.random.permutation(arr)
print(arr2)
```

Выберем случайно?

```python
folks = np.array(['John', 'Jane', 'Jack', 'Jill', 'Ann'])
chosen_person = np.random.choice(folks)
print(chosen_person)

#Выберем двоих, которых отправим в разведку
chosen_persons = np.random.choice(folks, size=2, replace=False)
print(chosen_persons)
```

Зададим seed!

```python
np.random.seed(23)
arr = np.random.randint(10, size=(3,4))
print(arr)

np.random.seed(42)
for i in range(3):
    print(np.random.rand())

print('то же самое :)')
np.random.seed(42)
for i in range(3):
    print(np.random.rand())
```
